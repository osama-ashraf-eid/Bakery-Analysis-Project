## Checking Columns in depth

### Checking Categorical Columns in depth 

```python
cat_cols=df.select_dtypes(include="object").columns
cat_cols
```

```python
for col in cat_cols:
    print(col)
    print(df[col].nunique())
    print(df[col].unique())
    print("-"*100)
```

### Check Numerical Columns in depth 

```python
num_cols=df.select_dtypes(include="number").columns
num_cols
```

```python
for col in num_cols:
    print(col)
    print(df[col].nunique())
    print(df[col].unique())
    print("-"*100)
```

```python
for col in num_cols:
    px.histogram(df, x=col, title=f"Distribution of {col}").show()
```
